```python
%%sql
drop table if exists regex;
CREATE TABLE regex(
    id INT AUTO_INCREMENT primary key,
    text VARCHAR(100)
);

INSERT INTO regex (text) VALUES 
('apple'),         
('Banana'),        
('cherry'),        
('date'),          
('elderberry'),    
('fig'),           
('grape'),         
('honeydew'),      
('running'),       
('123abc');        
```

```python
%%sql
select * from regex;
```

```python
%%sql
select * from regex where text regexp '^a';
```

```python
%%sql
select * from regex where text regexp 'e$';
```

```python
%%sql
select * from regex where text regexp '^[0-9]';
```
