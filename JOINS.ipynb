```python
%%sql
CREATE TABLE Students (
    student_id INT PRIMARY KEY,
    student_name VARCHAR(50),
    age INT
);

INSERT INTO Students VALUES
(1, 'Amit', 20),
(2, 'Priya', 22),
(3, 'Ravi', 21),
(4, 'Neha', 23);

CREATE TABLE Courses (
    course_id INT PRIMARY KEY,
    course_name VARCHAR(50),
    student_id INT
);

INSERT INTO Courses VALUES
(101, 'Maths', 1),
(102, 'Science', 2),
(103, 'History', 1),
(104, 'English', 3);
```

```python
%%sql
select s.student_id,s.student_name,c.course_name from Students s INNER JOIN Courses c on s.student_id=c.student_id;
```

```python
%%sql
select s.student_id,s.student_name,c.course_name from Students s left join Courses c on s.student_id=c.student_id;
```

```python
%%sql
CREATE TABLE customer (
    customer_id INT PRIMARY KEY,
    customer_name VARCHAR(50),
    city VARCHAR(50)
);

INSERT INTO customer VALUES
(1, 'Amit', 'Delhi'),
(2, 'Priya', 'Mumbai'),
(3, 'Ravi', 'Bangalore'),
(4, 'Neha', 'Chennai');
```

```python
%%sql
drop table orders;
CREATE TABLE Orders (
    order_id INT PRIMARY KEY,
    order_date DATE,
    amount int,
    customer_id INT
);

INSERT INTO Orders VALUES
(101, '2025-09-01', 1200.50, 1),
(102, '2025-09-05', 800.00, 2),
(103, '2025-09-07', 450.00, 1),
(104, '2025-09-10', 600.00, 3);
```

```python
%%sql
select c.customer_name,c.customer_id,o.amount,o.order_id from customer c left join orders o on c.customer_id=o.customer_id;
```

```python
%%sql
select c.city,c.customer_name,o.order_id as commonid from customer c inner join orders o on c.customer_id=o.customer_id;
```

```python
%%sql
select customer_id from customer union select order_id from orders;
```
